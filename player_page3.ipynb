```python
import pandas as pd
import plotly.express as px
df=pd.read_csv(r'C:\Users\ASUS\FootballPlayerAnalyticsProject\data\FC26_cleaned.csv')
df.info()
```

```python
# Page 3: Player Archetype & Physical Attribute Profiling

# 7. Sprint Speed vs. Strength Correlative Analysis
# Aggregate data (Mean Sprint Speed and Mean Strength)
agg_df = df.groupby(
    ['preferred_foot','league_name']
)[
    ['movement_sprint_speed','power_strength']
].mean().reset_index()


fig=px.scatter(
    agg_df,

    x='movement_sprint_speed',

    y="power_strength",

    color="preferred_foot",

    title='Average Sprint Speed vs Strength by Preferred Foot',

    color_discrete_sequence=["#3B528B", "#5DC863"],

    template='plotly_white',

    hover_data=[
        'league_name'
    ]

)


fig.update_traces(

    marker=dict(

        size=12,

        opacity=0.8,

        line=dict(

            color='black',

            width=0.5),

        symbol='circle'
    ),

    hovertemplate=

    "<b>%{customdata[0]}</b><br>" +

    "League: %{customdata[0]}<br>" +

    "Average Sprint Speed: %{x}<br>" +

    "Average Strength: %{y}<extra></extra>"

)


fig.update_layout(

    title=dict(

        text="Average Sprint Speed vs Strength by Preferred Foot",

        x=0.5,

        font=dict(size=24)

    ),

    xaxis_title="Average Sprint Speed",

    yaxis_title="Average Strength",

    legend_title="Preferred Foot",

    width=900,

    height=650,

    font=dict(

        family="Arial",

        size=14

    ),

    margin=dict(

        l=70,

        r=40,

        t=80,

        b=70

    )

)



fig.update_xaxes(

    showgrid=True,

    gridcolor="lightgray",

    showline=True,

    linecolor="black"

)



fig.update_yaxes(

    showgrid=True,

    gridcolor="lightgray",

    showline=True,

    linecolor="black"

)



fig.show()
# fig=px.scatter(df,
#                 x='movement_sprint_speed',
#                 y="power_strength", 
#                 color="preferred_foot",
#                 title='Sprint Speed vs Strength by Preferred Foot',
#                 color_discrete_sequence=["#3B528B", "#5DC863"],
#                 template='plotly_white',
#                 hover_data=[
#                     'short_name',
#                     'league_name',
#                     'club_name',
#                     'overall',
#                     'age'
#                 ]   
# )
# fig.update_traces(
#                  marker=dict(
#                     size=8,
#                     opacity=0.8,
#                     line=dict(
#                         color='black',
#                         width=0.5),
#                         symbol='circle'),
#                     hovertemplate=
#     "<b>%{customdata[0]}</b><br>" +
#     "Club: %{customdata[1]}<br>" +
#     "League: %{customdata[2]}<br>" +
#     "Sprint Speed: %{x}<br>" +
#     "Strength: %{y}<br>" +
#     "Overall: %{customdata[3]}<br>" +
#     "Age: %{customdata[4]}<extra></extra>"
# )
# fig.update_layout(
#     title=dict(
#         text="Sprint Speed vs Strength by Preferred Foot",
#         x=0.5,
#         font=dict(size=24)
#     ),
#     xaxis_title="Sprint Speed",
#     yaxis_title="Strength",
#     legend_title="Preferred Foot",
#     width=1000,
#     height=650,
#     font=dict(
#         family="Arial",
#         size=14
#     ),
#     margin=dict(
#         l=70,
#         r=40,
#         t=80,
#         b=70
#     )
# )

# fig.update_xaxes(
#     showgrid=True,
#     gridcolor="lightgray",
#     showline=True,
#     linecolor="black"
# )

# fig.update_yaxes(
#     showgrid=True,
#     gridcolor="lightgray",
#     showline=True,
#     linecolor="black"
# )

# fig.show()

```

```python
# 8. Age Spread Patterns by Dominant Foot -if u want to show names as well u can use scatter plot 
fig=px.histogram(df,
                 x="age", 
                 color="preferred_foot", 
                 barmode="overlay",
                 nbins=25,
                 opacity=0.85,
                 title='Age Distribution by Preferred Foot',
                 template='plotly_white',
                 color_discrete_sequence=["#3B528B", "#5EC962"],
                 hover_data=[
                     'short_name',
                     'club_name',
                     'overall'  
                 ],
)
fig.update_traces(
    hovertemplate=
    '<b>Age:</b>%{x}<br>'+
    '<b>Players:</b>%{y}<br>'+
    '<b>Preferred Foot:</b>%{%{fullData.name}<extra></extra>',

marker=dict(
        line=dict(
            color="black",
            width=1
        )
    )
)
fig.update_layout(
    title=dict(
        text='Age Distribution by Preferred Foot',
        font=dict(size=24),
        x=0.5
    ),
    xaxis_title='Age (Years)',
    yaxis_title='Number of Players',
    legend_title='Preferred Foot',
    height=600,
    width=950,
    font=dict(
        family='Arial',
        size=14
    ),
     margin=dict(
        l=70,
        r=40,
        t=80,
        b=70
    ),
    bargap=0.05
)

fig.update_xaxes(
    showgrid=False,
    showline=True,
    linecolor="black"
)

fig.update_yaxes(
    showgrid=True,
    gridcolor="lightgray",
    showline=True,
    linecolor="black"
)
fig.show()
```

```python
# 9. Physicality Ratings Based on Preferred Foot [violin->(kde+box)]
fig=px.violin(
    df,
     x="preferred_foot", 
     y="physic", 
     color="preferred_foot",
     box=True,
     color_discrete_sequence=["#3B528B", "#5EC962"],
     points='outliers',
     template='plotly_white',
     title='Physical Rating Distribution by Preferred Foot',
     hover_data=[
        "short_name",
        "club_name",
        "league_name",
        "overall",
        "age"
     ]
)
fig.update_traces(
    meanline_visible=True,
    hovertemplate=
    "<b>%{customdata[0]}</b><br>" +
    "Club: %{customdata[1]}<br>" +
    "League: %{customdata[2]}<br>" +
    "Overall: %{customdata[3]}<br>" +
    "Age: %{customdata[4]}<br>" +
    "Physic: %{y}<br>" +
    "Preferred Foot: %{x}<extra></extra>"
)
fig.update_layout(
    title=dict(
        text="Physical Rating Distribution by Preferred Foot",
        x=0.5,
        font=dict(
            size=24,
            family="Arial"
        )
    ),
    xaxis_title="Preferred Foot",
    yaxis_title="Physical Rating",
    legend_title="Preferred Foot",
    width=950,
    height=600,
    font=dict(
        family="Arial",
        size=14
    ),
    margin=dict(
        l=70,
        r=40,
        t=80,
        b=70
    )
)

fig.update_xaxes(
    showgrid=False,
    showline=True,
    linecolor="black"
)

fig.update_yaxes(
    showgrid=True,
    gridcolor="lightgray",
    showline=True,
    linecolor="black",
    range=[0, 100]
)
fig.show()
```

```python
# 10. Positional Hierarchies and Dominant Foot Layouts-Visually groups player values based on regional leagues and preferred foot combinations to display asset concentrations.
fig = px.treemap(
    df,
    path=["league_name", "preferred_foot"],
    values="value_eur",
    color="value_eur",
    color_continuous_scale="Viridis",
    template="plotly_white",
    title="Player Market Value Distribution by League and Preferred Foot"
)

fig.update_traces(
    textinfo="label+percent parent",
    hovertemplate=
    "<b>%{label}</b><br>" +
    "Market Value: €%{value:,.0f}<br>" +
    "Share in Parent: %{percentParent}<br>" +
    "Share Overall: %{percentRoot}<extra></extra>"
)

fig.update_layout(
    title=dict(
        text="Player Market Value Distribution by League and Preferred Foot",
        x=0.5,
        font=dict(size=24)
    ),
    width=950,
    height=700,
    margin=dict(l=20, r=20, t=80, b=20),
    font=dict(family="Arial", size=14),
    coloraxis_colorbar=dict(title="Market Value (€)")
)

fig.show()
```
