```python
# Page 2: League & Club Structural Evaluations
import pandas as pd
import plotly.express as px
```

```python
df=pd.read_csv(r'C:\Users\ASUS\FootballPlayerAnalyticsProject\data\FC26_cleaned.csv')
df.info()
```

```python
# 4. Player Technical Skill Spreads Across Top 5 Leagues
df['league_name'].unique()
```

```python
top5=['Premier League','La Liga','Serie A','Bundesliga','Ligue 1']
df_top5=df[df['league_name'].isin(top5)]
print(df_top5["league_name"].unique())
```

```python
fig=px.violin(df_top5,
              x='league_name',
              y='dribbling',
              color='league_name',
              box=True,
              points='outliers',
              template='plotly_white',
              title="Player Technical Skill Spreads Across Top 5 Leagues- Dribbling",
              color_discrete_sequence=px.colors.qualitative.Set2,
              hover_data=['short_name','club_name','age']
              )
fig.update_traces(
    meanline_visible=True,
    hovertemplate=
        '<b>League:</b>%{x}<br>'+
        '<b>Dribbling:</b>{y}<br>'+
        '<b>Player:</b>%{customdata[0]}<br>'+
        "<b>Club:</b> %{customdata[1]}<br>" +
        "<b>Age:</b> %{customdata[2]}<extra></extra>"
)
fig.update_layout(
    title=dict(
        text="Player Technical Skill Spreads Across Top 5 European Leagues-Dribbling",
        x=0.5,
        font=dict(size=24)
    ),

    xaxis_title="League",
    yaxis_title="Dribbling Rating",

    width=1000,
    height=600,

    font=dict(
        family="Arial",
        size=14
    ),

    margin=dict(
        l=70,
        r=40,
        t=80,
        b=70
    ),

    legend_title="League"
)

fig.update_xaxes(
    showgrid=False,
    showline=True,
    linewidth=1,
    linecolor="black",
)

fig.update_yaxes(
    showgrid=True,
    gridcolor="lightgray",
    zeroline=False,
    showline=True,
    linewidth=1,
    linecolor="black",
)


fig.show()
```

```python
# all skills by grouped bar chart
skill_avg=df_top5.groupby('league_name')[["pace", "shooting", "passing", "dribbling", "defending", "physic"]].mean().reset_index()
```

```python
skill_avg
```

```python
# conversion from wide format to long format 
skill_avg=skill_avg.melt(
    id_vars="league_name",
    var_name="Technical Skill",
    value_name="Average Rating"
)
```

```python
# 5. Squad Rating Variances within Leagues-It compares how player overall ratings are distributed within each league.
fig=px.box(
    df_top5,
    x='league_name',
    y='overall',
    color='league_name',
    template='plotly_white',
    title='Squad Rating Variances within Top 5 European Leagues',
    points="outliers",
    color_discrete_sequence=px.colors.qualitative.Dark2,
    hover_data=['short_name','club_name','age','overall']
)
fig.update_traces(
    hovertemplate=
    '<b>%{customdata[0]}</b><br>'+ # custom data contains columns you specified in hover_data
    '<b>Club:%{customdata[1]}<br>'+
    "League: %{x}<br>" +
    "Overall Rating: %{y}<br>" +
    "Age: %{customdata[2]}<extra></extra>",
    boxmean=True,
    whiskerwidth=0.8,
    marker=dict(
        size=5,
        opacity=0.8,
        line=dict(
            color='black',
            width=0.5
        )

    )
)
fig.update_layout(
    title=dict(
        text='Squad Rating Variances within Top 5 European Leagues',
        x=0.5,
        font=dict(
            family='Arial',
            size=24,
            color="#2F2F2F"
        )
    ),
    xaxis_title='League',
    yaxis_title='Overall Player Rating',
    font=dict(
        family='Arial',
        size=14
    ),
    height=600,
    width=1000,
    legend_title="League",

    margin=dict(
        l=70,
        r=40,
        t=80,
        b=70
    ),

    hovermode="closest"
)
fig.update_xaxes(
    showgrid=False,
    showline=True,
    linecolor='black',
    linewidth=1
)
fig.update_yaxes(
    showgrid=True,
    gridcolor="lightgray",
    gridwidth=1,
    zeroline=False,
    showline=True,
    linewidth=1,
    linecolor="black",
    range=[40, 100]
)
fig.show()
```

```python
# 6. League Financial Wage Mix Architecture

# top 5 european leagues
top5 = [
    'La Liga', 'Bundesliga', 'Ligue 1', 'Serie A', 'Premier League'
]
#top 5 leagues filtered
dftop5=df[df['league_name'].isin(top5)]

# sum of wages 
df_sum=dftop5.groupby(['league_name','club_name'])['wage_eur'].sum().reset_index()

# top 10 clubs 
df_sum = (
    df_sum.sort_values("wage_eur", ascending=False)
            .groupby("league_name")
            .head(10)
)
```

```python

fig = px.treemap(
    df_sum,
    path=["league_name", "club_name"],
    values="wage_eur",
    color="wage_eur",
    color_continuous_scale="Viridis",
    template="plotly_white",
    title="League Financial Wage Mix Architecture",
    hover_data={"wage_eur": ":,"}
)

fig.update_traces(
    textinfo="label+value+percent parent",
    hovertemplate=
    "<b>%{label}</b><br>" +
    "Total Wage: €%{value:,.0f}<br>" +
    "League Share: %{percentParent}<br>" +
    "Overall Share: %{percentRoot}<extra></extra>",

    marker=dict(
        line=dict(
            color="white",
            width=2
        )
    )
)

fig.update_layout(
    title=dict(
        text="League Financial Wage Mix Architecture",
        x=0.5,
        font=dict(
            size=24,
            family="Arial",
            color="#2F2F2F"
        )
    ),

    width=1100,
    height=700,

    margin=dict(
        l=20,
        r=20,
        t=80,
        b=20
    ),

    font=dict(
        family="Arial",
        size=14
    ),

    coloraxis_colorbar=dict(
        title="Total Wage (€)"
    )
)

fig.show()
```
